# Arquitectura De La Red Neuronal | GraphSAGE 📐📏🧱🏗️

````python
import networkx as nx
import matplotlib.pyplot as plt

# Crear grafo de muestreo (k=2)
G = nx.DiGraph()
G.add_edges_from([
    ("v", "u1"), ("v", "u2"), ("v", "u3"),  # Capa 1
    ("u1", "w1"), ("u1", "w2"),             # Capa 2 (hijos de u1)
    ("u2", "w3"), ("u2", "w4"),             # Capa 2 (hijos de u2)
    ("u3", "w5")                             # Capa 2 (hijos de u3)
])

# Posiciones jerárquicas
pos = {
    "v": (0, 0),
    "u1": (-1, 1), "u2": (0, 1), "u3": (1, 1),
    "w1": (-1.5, 2), "w2": (-0.5, 2), "w3": (0.5, 2), 
    "w4": (1.5, 2), "w5": (1, 2)
}

# Dibujar nodos y aristas
nx.draw_networkx_nodes(G, pos, nodelist=["v"], node_color="green", node_size=500)
nx.draw_networkx_nodes(G, pos, nodelist=["u1", "u2", "u3"], node_color="blue", node_size=400)
nx.draw_networkx_nodes(G, pos, nodelist=["w1", "w2", "w3", "w4", "w5"], node_color="red", node_size=300)
nx.draw_networkx_edges(G, pos, arrowstyle="->", arrowsize=15)

# Etiquetas y agregación
plt.text(0, -0.5, "Nodo Objetivo (v)", ha="center")
plt.text(-1, 1.5, "AGGREGATE", fontsize=8, bbox=dict(facecolor="white", alpha=0.7))
plt.title("GraphSAGE: Muestreo y Agregación (k=2)")
plt.show()
````

# Construcción Grafo

## Descripción General

Este código inicializa y construye un grafo no dirigido utilizando la biblioteca **NetworkX**, donde los nodos representan emprendimientos y sus atributos contienen información relevante extraída de una base de datos relacional.

---

### Estructura del Código

**`nx.Graph()`**  
Crea un objeto de grafo no dirigido (las aristas no tienen dirección) utilizando la biblioteca NetworkX.  
Este tipo de grafo es apropiado porque las relaciones entre emprendimientos (que se añadirán posteriormente) son **simétricas**.

---

**`for _, row in emprendimientos.iterrows():`**  
Itera sobre cada fila del DataFrame `emprendimientos` que contiene la información de cada emprendimiento.

- `_`: Ignora el índice de la fila.
- `row`: Contiene los datos del emprendimiento actual.

---

**`temas = emprendimiento_tematica[emprendimiento_tematica['id_emprendimiento'] == row['id_emprendimiento']]['id_tematica'].tolist()`**  
Consulta la tabla de relación muchos-a-muchos `emprendimiento_tematica` para obtener las temáticas asociadas al emprendimiento actual.

- Filtra las filas donde `id_emprendimiento` coincide con el actual.
- Extrae los valores de `id_tematica`.
- Convierte el resultado a una lista de temáticas (categorías) asociadas.

---

**Atributos del nodo**

```python
node_attrs = {
    'nombre_emprendimiento': row['nombre_emprendimiento'],
    'descripcion': row['descripcion'] if pd.notna(row['descripcion']) else '',
    'redes_sociales': row['redes_sociales'] if pd.notna(row['redes_sociales']) else '',
    'sitio_web': row['sitio_web'],
    'id_municipio_origen': row['id_municipio_origen'],
    'id_alcance': row['id_alcance'],
    'tematicas': temas
}
```

Crea un diccionario con los atributos del nodo:

- **`nombre_emprendimiento`**: Nombre del emprendimiento (cadena).
- **`descripcion`**: Descripción del emprendimiento (cadena), con manejo de valores nulos.
- **`redes_sociales`**: Redes sociales (cadena), con manejo de nulos.
- **`sitio_web`**: URL del sitio web (cadena).
- **`id_municipio_origen`**: ID del municipio de origen (entero).
- **`id_alcance`**: ID del alcance (entero).
- **`tematicas`**: Lista de IDs de temáticas asociadas (lista de enteros).

-------

**`G.add_node(row['id_emprendimiento'], **node_attrs)`** 

Añade un nodo al grafo G con:

    Identificador: row['id_emprendimiento'] (clave primaria de la tabla Emprendimiento).

    Atributos: Todos los definidos en node_attrs mediante ** que desempaqueta el diccionario.

````python
# Inicializar el grafo
G = nx.Graph()

# Agregar nodos (emprendimientos)
for _, row in emprendimientos.iterrows():
    # Obtener temáticas asociadas
    temas = emprendimiento_tematica[emprendimiento_tematica['id_emprendimiento'] == row['id_emprendimiento']]['id_tematica'].tolist()
    # Atributos del nodo
    node_attrs = {
        'nombre_emprendimiento': row['nombre_emprendimiento'],
        'descripcion': row['descripcion'] if pd.notna(row['descripcion']) else '',
        'redes_sociales': row['redes_sociales'] if pd.notna(row['redes_sociales']) else '',
        'sitio_web': row['sitio_web'],
        'id_municipio_origen': row['id_municipio_origen'],
        'id_alcance': row['id_alcance'],
        'tematicas': temas
    }
    G.add_node(row['id_emprendimiento'], **node_attrs)

print(f"Nodos creados: {G.number_of_nodes()}") # Imprime el número de nodos
````

# Creación de Aristas para el Grafo 

Este código establece las conexiones (aristas) entre los emprendimientos (nodos) del grafo, considerando tres tipos de relaciones:
1. **Interacciones** basadas en publicaciones y temáticas comunes
2. **Similitudes** por características compartidas (temáticas, municipio, alcance)
3. **Influencia** medida por cantidad de seguidores

-----

## Parte 1: Aristas por Interacciones (Publicaciones)

Se tiene lo siguiente:

*`publicaciones`*: un DataFrame con publicaciones, donde cada publicación está asociada a un id_emprendimiento.

*`emprendimiento_tematica`*: un DataFrame que relaciona emprendimientos con temáticas. Es una tabla de relación muchos-a-muchos entre emprendimientos y temáticas.

La idea es si dos emprendimientos comparten una misma temática, se consideran relacionados, aunque no haya información directa de interacción.



```python
for _, pub in publicaciones.iterrows():
    id_emprendimiento = pub['id_emprendimiento']
    # Conectar emprendimientos que comparten temáticas en publicaciones
    related_emprendimientos = emprendimiento_tematica[
        emprendimiento_tematica['id_tematica'].isin(
            emprendimiento_tematica[emprendimiento_tematica['id_emprendimiento'] == id_emprendimiento]['id_tematica']
        )
    ]['id_emprendimiento'].unique()
```

### Explicación:
1. **Flujo**: Itera sobre cada publicación en el DataFrame `publicaciones`.
2. **Lógica de conexión**:
   - Encuentra todos los emprendimientos (`related_emprendimientos`) que comparten al menos una temática con el emprendimiento que hizo la publicación.
   - Usa la tabla `emprendimiento_tematica` para esta relación muchos-a-muchos.
   - `unique()` asegura no tener duplicados.

```python
    for rel_emp in related_emprendimientos:
        if rel_emp != id_emprendimiento and rel_emp in G.nodes:
            weight = pub['n_likes'] if pd.notna(pub['n_likes']) else 0
            G.add_edge(id_emprendimiento, rel_emp, weight=weight + 1)
```

### Explicación:
1. **Creación de aristas**:
   - Para cada emprendimiento relacionado, crea una arista si:
     - No es el mismo emprendimiento (`rel_emp != id_emprendimiento`)
     - Existe en el grafo (`rel_emp in G.nodes`)
   - **Peso**: Basado en los "likes" de la publicación (`n_likes`), con:
     - Manejo de valores nulos (default 0)
     - +1 para evitar pesos cero (suavizado)

> **Propósito**: Modela que los emprendimientos que interactúan con temáticas similares están relacionados, donde la popularidad de la publicación (likes) afecta la fuerza de la conexión.
>
> Si dos emprendimientos comparten temáticas y uno publica algo que recibe muchos likes → hay una señal de relevancia o afinidad temática.
> El grafo resultante podría reflejar cuán influyente o visible es un emprendimiento en su grupo temático.

-----

## Parte 2: Aristas por Similitudes

```python
for emp1 in G.nodes:
    for emp2 in G.nodes:
        if emp1 < emp2:  # Evitar duplicados
```

### Explicación:
- Compara todos los pares de nodos únicos (`emp1 < emp2` evita comparar A-B y B-A).

```python
            emp1_attrs = G.nodes[emp1]
            emp2_attrs = G.nodes[emp2]
            # Similitud por temática
            temas_comunes = len(set(emp1_attrs['tematicas']) & set(emp2_attrs['tematicas']))
            # Similitud por municipio y alcance
            same_municipio = 1 if emp1_attrs['id_municipio_origen'] == emp2_attrs['id_municipio_origen'] else 0
            same_alcance = 1 if emp1_attrs['id_alcance'] == emp2_attrs['id_alcance'] else 0
```

### Explicación:
1. **Métricas de similitud**:
   - **Temáticas comunes**: Cuenta intersección de listas de temáticas.
   - **Municipio**: 1 si son del mismo municipio, 0 si no.
   - **Alcance**: 1 si tienen mismo alcance (local/regional).

```python
            # Ponderar arista
            weight = temas_comunes + same_municipio + same_alcance
            if weight > 0:
                G.add_edge(emp1, emp2, weight=weight)
```

### Explicación:
- **Peso total**: Suma de las tres métricas (puede ser 0-3).
- Solo crea arista si hay al menos una similitud (`weight > 0`).

> **Propósito**: Conecta emprendimientos con características intrínsecas similares, donde más coincidencias generan conexiones más fuertes.


-----

## Parte 3: Ajuste por Seguidores

```python
for _, row in seguidores.iterrows():
    id_emprendimiento = row['id_emprendimiento']
    if id_emprendimiento in G.nodes:
        G.nodes[id_emprendimiento]['seguidores'] = row['cantidad']
```

### Explicación:
- Añade atributo `seguidores` a cada nodo con el valor de la tabla `seguidores`.

```python
        # Aumentar peso de aristas existentes según seguidores
        for emp2 in G.neighbors(id_emprendimiento):
            G[id_emprendimiento][emp2]['weight'] += row['cantidad'] / 1000  # Normalizar
```

### Explicación:
- **Refuerzo de aristas**: Para cada conexión existente del emprendimiento:
  - Aumenta el peso proporcional a sus seguidores.
  - Divide entre 1000 para normalizar (evitar dominancia numérica).

> **Propósito**: Modela que emprendimientos con más seguidores influyen más en sus conexiones.


-----

## Resumen Final
```python
print(f"Aristas creadas: {G.number_of_edges()}")
```
- Reporta el total de aristas creadas combinando los tres métodos.

## Estrategia General
1. **Interacciones** → Datos observables (publicaciones)
2. **Similitudes** → Características intrínsecas
3. **Seguidores** → Popularidad/influencia

Esta combinación permite que GraphSage aprenda tanto de:
- Comportamiento real (interacciones)
- Atributos estructurales (similitudes)
- Señales de influencia social (seguidores)

> **Nota**: Los pesos son acumulativos (se suman cuando múltiples criterios aplican), lo que permite conexiones más ricas.

-----

````python
# Parte 1: Crear aristas basadas en interacciones

for _, pub in publicaciones.iterrows():
    id_emprendimiento = pub['id_emprendimiento']
    # Conectar emprendimientos que comparten temáticas en publicaciones (simplificación, ya que no tenemos datos de quién comentó)
    related_emprendimientos = emprendimiento_tematica[
        emprendimiento_tematica['id_tematica'].isin(
            emprendimiento_tematica[emprendimiento_tematica['id_emprendimiento'] == id_emprendimiento]['id_tematica']
        )
    ]['id_emprendimiento'].unique()
    # Excluir el emprendimiento actual de la lista de relacionados
    related_emprendimientos = related_emprendimientos[related_emprendimientos != id_emprendimiento]
 
    for rel_emp in related_emprendimientos:
        if rel_emp != id_emprendimiento and rel_emp in G.nodes:
            weight = pub['n_likes'] if pd.notna(pub['n_likes']) else 0
            G.add_edge(id_emprendimiento, rel_emp, weight=weight + 1)
````

````python
# Parte 2: Crear aristas basadas en similitudes

for emp1 in G.nodes:
    for emp2 in G.nodes:
        if emp1 < emp2:  # Evitar duplicados
            emp1_attrs = G.nodes[emp1]
            emp2_attrs = G.nodes[emp2]
            # Similitud por temática
            temas_comunes = len(set(emp1_attrs['tematicas']) & set(emp2_attrs['tematicas']))
            # Similitud por municipio y alcance
            same_municipio = 1 if emp1_attrs['id_municipio_origen'] == emp2_attrs['id_municipio_origen'] else 0
            same_alcance = 1 if emp1_attrs['id_alcance'] == emp2_attrs['id_alcance'] else 0
            # Ponderar arista
            weight = temas_comunes + same_municipio + same_alcance
            if weight > 0:
                G.add_edge(emp1, emp2, weight=weight)
````

````python
# Parte 3: Agregar peso basado en seguidores

for _, row in seguidores.iterrows():
    id_emprendimiento = row['id_emprendimiento']
    if id_emprendimiento in G.nodes:
        G.nodes[id_emprendimiento]['seguidores'] = row['cantidad']
        # Aumentar peso de aristas existentes según seguidores
        for emp2 in G.neighbors(id_emprendimiento):
            G[id_emprendimiento][emp2]['weight'] += row['cantidad'] / 1000  # Normalizar

print(f"Aristas creadas: {G.number_of_edges()}")
````

## Objetivo del código

Crear un **vector de características numéricas y categóricas normalizadas y codificadas** para cada emprendimiento, y asignarlas a los nodos del grafo `G`.

La siguiente celda construye y agrega vectores de características (`features`) para cada nodo del grafo `G`, utilizando datos numéricos, categóricos y multiclase (temáticas). Este proceso es esencial para usar modelos de aprendizaje como **GraphSAGE**.

---

##  Paso a paso 

---

### 1. **Normalización de atributos numéricos**

```python
scaler = MinMaxScaler()
```

* Se crea un **escalador de Min-Max** (de `sklearn.preprocessing`) para que los valores numéricos estén entre 0 y 1.
* Esto es importante porque redes neuronales funcionan mejor con datos normalizados.

---

### 2. **Cálculo de likes por emprendimiento**

```python
likes_por_emprendimiento = publicaciones.groupby('id_emprendimiento')['n_likes'].sum().reset_index()
likes_por_emprendimiento.columns = ['id_emprendimiento', 'total_likes']
```

* Agrupa las publicaciones por `id_emprendimiento` y **suma la cantidad de likes**.
* Resultado: un DataFrame con la cantidad total de likes por cada emprendimiento.

#### Resultado:

| id\_emprendimiento | total\_likes |
| ------------------ | ------------ |
| A                  | 45           |
| B                  | 78           |

---

### 3. **Combinar likes y seguidores con emprendimientos**

```python
emprendimientos_features = emprendimientos.merge(likes_por_emprendimiento, on='id_emprendimiento', how='left').fillna({'total_likes': 0})
emprendimientos_features = emprendimientos_features.merge(seguidores[['id_emprendimiento', 'cantidad']], on='id_emprendimiento', how='left').fillna({'cantidad': 0})
```

* Se integran las columnas `total_likes` y `cantidad` (número de seguidores) al DataFrame principal de emprendimientos.
* Se llenan los valores nulos con 0 (por ejemplo, si un emprendimiento no tiene publicaciones ni seguidores).

---

### 4. **Normalización de variables numéricas**

```python
numeric_features = scaler.fit_transform(emprendimientos_features[['total_likes', 'cantidad']])
emprendimientos_features[['total_likes_norm', 'cantidad_norm']] = numeric_features
```

* Normaliza las columnas `total_likes` y `cantidad` entre 0 y 1.
* Se agregan dos nuevas columnas: `total_likes_norm` y `cantidad_norm`.

---

### 5. **Codificación one-hot del municipio de origen**

```python
onehot_encoder_municipio = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
municipio_encoded = onehot_encoder_municipio.fit_transform(emprendimientos[['id_municipio_origen']])
municipio_encoded_df = pd.DataFrame(municipio_encoded, columns=onehot_encoder_municipio.get_feature_names_out(['id_municipio_origen']))
```

* Convierte la variable `id_municipio_origen` en una representación **one-hot** (binaria).
* Cada municipio genera una columna.

#### Ejemplo:

| id\_municipio\_origen | id\_municipio\_origen\_1 | id\_municipio\_origen\_2 | id\_municipio\_origen\_3 |
| --------------------- | ------------------------ | ------------------------ | ------------------------ |
| 1                     | 0                        | 1.0                      | 0                        |
| 2                     | 1.0                      | 0                        | 0                        |
| 3                     | 0                        | 0                        | 1.0                      |

---

### 6. **Codificación one-hot del alcance**

```python
onehot_encoder_alcance = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
alcance_encoded = onehot_encoder_alcance.fit_transform(emprendimientos[['id_alcance']])
alcance_encoded_df = pd.DataFrame(alcance_encoded, columns=onehot_encoder_alcance.get_feature_names_out(['id_alcance']))
```

* Igual que lo anterior, pero para el campo `id_alcance`.

---

### 7. **Codificación binaria de temáticas por emprendimiento**

```python
tematica_encoded = np.zeros((len(emprendimientos), len(tematicas)))
for _, row in emprendimiento_tematica.iterrows():
    idx_emp = emprendimientos.index[emprendimientos['id_emprendimiento'] == row['id_emprendimiento']].tolist()[0]
    idx_tem = tematicas.index[tematicas['id_tematica'] == row['id_tematica']].tolist()[0]
    tematica_encoded[idx_emp, idx_tem] = 1
```

* Crea una **matriz binaria** donde:

  * Las filas representan emprendimientos
  * Las columnas representan temáticas
  * Un 1 indica que el emprendimiento está asociado a esa temática

#### Ejemplo:

| tematica\_1 | tematica\_2 | tematica\_3 |
| ----------- | ----------- | ----------- |
| 1           | 0           | 1           |

---

### 8. **Construcción del DataFrame final con características**

```python
features_df = pd.concat([
    emprendimientos_features[['id_emprendimiento', 'total_likes_norm', 'cantidad_norm']],
    municipio_encoded_df,
    alcance_encoded_df,
    tematica_encoded_df
], axis=1)
```

* Une:

  * Atributos numéricos normalizados (`total_likes_norm`, `cantidad_norm`)
  * One-hot de municipio
  * One-hot de alcance
  * Binarios de temáticas

---

### 9. **Asignación de características al grafo `G`**

```python
for _, row in features_df.iterrows():
    if row['id_emprendimiento'] in G.nodes:
        G.nodes[row['id_emprendimiento']]['features'] = row.drop('id_emprendimiento').values
```

* Recorre el `features_df` y asigna el vector de características al nodo correspondiente del grafo `G`.

---

### 10. **Impresión del tamaño de las características**

```python
print("Dimensiones de las características:\n", features_df.shape)
```

* Muestra cuántos emprendimientos (`n`) y cuántas características (`m`) se generaron.
* Esto será útil para verificar la **dimensión del input** que irá a la red neuronal.

* **Escalado y codificado**: facilita el aprendizaje automático.
* **Flexible**: puedes agregar más atributos numéricos o categóricos.



````python
# Normalizar atributos numéricos
scaler = MinMaxScaler()

# Calcular total de likes por emprendimiento
likes_por_emprendimiento = publicaciones.groupby('id_emprendimiento')['n_likes'].sum().reset_index()
likes_por_emprendimiento.columns = ['id_emprendimiento', 'total_likes']

# Combinar likes y seguidores
emprendimientos_features = emprendimientos.merge(likes_por_emprendimiento, on='id_emprendimiento', how='left').fillna({'total_likes': 0})
emprendimientos_features = emprendimientos_features.merge(seguidores[['id_emprendimiento', 'cantidad']], on='id_emprendimiento', how='left').fillna({'cantidad': 0})

# Normalizar total_likes y cantidad
numeric_features = scaler.fit_transform(emprendimientos_features[['total_likes', 'cantidad']])
emprendimientos_features[['total_likes_norm', 'cantidad_norm']] = numeric_features

# Codificar atributos categóricos
onehot_encoder_municipio = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
municipio_encoded = onehot_encoder_municipio.fit_transform(emprendimientos[['id_municipio_origen']])
municipio_encoded_df = pd.DataFrame(municipio_encoded, columns=onehot_encoder_municipio.get_feature_names_out(['id_municipio_origen']))

onehot_encoder_alcance = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
alcance_encoded = onehot_encoder_alcance.fit_transform(emprendimientos[['id_alcance']])
alcance_encoded_df = pd.DataFrame(alcance_encoded, columns=onehot_encoder_alcance.get_feature_names_out(['id_alcance']))

# Codificar temáticas (matriz binaria)
tematica_encoded = np.zeros((len(emprendimientos), len(tematicas)))
for _, row in emprendimiento_tematica.iterrows():
    idx_emp = emprendimientos.index[emprendimientos['id_emprendimiento'] == row['id_emprendimiento']].tolist()[0]
    idx_tem = tematicas.index[tematicas['id_tematica'] == row['id_tematica']].tolist()[0]
    tematica_encoded[idx_emp, idx_tem] = 1
tematica_encoded_df = pd.DataFrame(tematica_encoded, columns=[f'tematica_{i}' for i in tematicas['id_tematica']])

# Combinar características numéricas y categóricas
features_df = pd.concat([emprendimientos_features[['id_emprendimiento', 'total_likes_norm', 'cantidad_norm']],
                         municipio_encoded_df, alcance_encoded_df, tematica_encoded_df], axis=1)

# Agregar características al grafo
for _, row in features_df.iterrows():
    if row['id_emprendimiento'] in G.nodes:
        G.nodes[row['id_emprendimiento']]['features'] = row.drop('id_emprendimiento').values

print("Dimensiones de las características:\n", features_df.shape)
````

# 📜Procesar Texto✍️

---

## 🧠 ¿Qué es BETO?

**BETO** es un modelo BERT preentrenado para **español**, desarrollado por el equipo de la Universidad Católica de Chile. Es similar a BERT-base en arquitectura, pero entrenado exclusivamente con texto en español.

El modelo que se está cargando es:

* 📦 Nombre en Hugging Face: `dccuchile/bert-base-spanish-wwm-uncased`
* 📌 Características:

  * Basado en BERT-base
  * Preentrenado con Whole Word Masking (WWM)
  * No distingue entre mayúsculas y minúsculas (`uncased`)

---

## 📄 Código

### 1. `from transformers import AutoTokenizer, AutoModel`

* Importa dos clases de la biblioteca `transformers`:

  * `AutoTokenizer`: se encarga de convertir texto en **tokens numéricos** que el modelo puede entender.
  * `AutoModel`: carga la **arquitectura del modelo preentrenado**, sin ninguna cabeza de tarea específica (por ejemplo, sin clasificación, sin traducción).

---

### 2. `tokenizer = AutoTokenizer.from_pretrained(...)`

* Descarga y carga automáticamente el **tokenizador** asociado al modelo BETO.
* El tokenizador:

  * Divide el texto en subtokens (WordPiece)
  * Añade tokens especiales (`[CLS]`, `[SEP]`, etc.)
  * Genera tensores listos para ser procesados por el modelo

#### 🧾 Ejemplo:

```python
inputs = tokenizer("Hola, ¿cómo estás?", return_tensors="pt")
```

Resultado:

```python
{
  'input_ids': tensor([[CLS, 15412, 117, 12236, 7949, 106, SEP]]),
  'attention_mask': tensor([[1, 1, 1, 1, 1, 1, 1]])
}
```

---

### 3. `model = AutoModel.from_pretrained(...)`

* Carga el **modelo BETO preentrenado** (solo la base de BERT, sin una cabeza específica).
* Devuelve una instancia de `BertModel`, que al ejecutarse sobre entradas tokenizadas devuelve los **embeddings contextuales** para cada token.

#### 🧾 Ejemplo:

```python
outputs = model(**inputs)
```

* `outputs.last_hidden_state` → tensores `[batch_size, seq_length, hidden_size]` (por defecto `hidden_size=768`).
* Cada vector representa el significado contextual del token.

---

## 📦 Finalidad del código

Después de ejecutarlo:

* Se puede usar BETO para tareas como:

  * Extracción de características (embeddings)
  * Fine-tuning para clasificación, NER, etc.
  * Comparación semántica de oraciones
* Hay acceso a una **representación densa del texto en español**, entrenada con millones de oraciones.

---

## ✅ Ventajas de usar `AutoTokenizer` y `AutoModel`

* Carga automáticamente el modelo correcto según el nombre (`dccuchile/bert-base-spanish-wwm-uncased`)
* No hay que preocuparse por la arquitectura específica
* Compatible con versiones actualizadas de la librería 🤗 Transformers

---

````python
# Cargar modelo BETO
tokenizer = AutoTokenizer.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
model = AutoModel.from_pretrained("dccuchile/bert-base-spanish-wwm-uncased")
````

---

###  Función: `get_text_embedding(text)`

#### 📌 Propósito:

Genera un **vector de embedding** de tamaño `768` para un texto en español, usando el modelo **BETO** (`bert-base-spanish-wwm-uncased`).

---

```python
def get_text_embedding(text):
```

* Define una función que recibe un `text` (cadena en español).

---

```python
    if not text or pd.isna(text):
        return np.zeros(768)  # Dimensión de BETO
```

* Si el texto está vacío (`""`, `None`) o es `NaN` (de un `DataFrame`), devuelve un vector de ceros de 768 dimensiones (para mantener compatibilidad dimensional).

---

```python
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=128)
```

* Usa el `tokenizer` de BETO para convertir el texto en tensores:

  * `return_tensors="pt"` → salida como tensores de PyTorch
  * `truncation=True` → corta textos largos
  * `padding=True` → añade padding si es necesario
  * `max_length=128` → longitud máxima del texto tokenizado

---

```python
    with torch.no_grad():
        outputs = model(**inputs)
```

* Desactiva el cálculo de gradientes (más eficiente) y obtiene las salidas del modelo BETO.
* `outputs.last_hidden_state` es de forma `[1, seq_len, 768]`.

---

```python
    return outputs.last_hidden_state.mean(dim=1).squeeze().numpy()
```

* Calcula el promedio de los embeddings de todos los tokens → un vector de tamaño `768`.
* Lo convierte a un arreglo de NumPy para su posterior uso.

---

### 🧾 Resultado:

Devuelve un **embedding denso y contextualizado** de la oración, que puedes usar como entrada en modelos, clustering, visualización, etc.

---

````python
# Función para obtener embeddings de texto
def get_text_embedding(text):
    if not text or pd.isna(text):
        return np.zeros(768)  # Dimensión de BETO
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=512)  # Mantener 512 tokens que es el maximo para no cortar texto
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1).squeeze().numpy()
````

# Integración de Texto y Embeddings para GraphSAGE

````python
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from scipy.sparse import csr_matrix
````

````python
# Cargar datos originales para mapeo
publicaciones = pd.read_csv('.\\DATOS\\publicaciones.csv')
comentarios = pd.read_csv('.\\DATOS\\comentarios.csv')

# 1. Procesamiento de embeddings textuales

def aggregate_text_embeddings(embeddings, ids, aggregation='mean'):
    """Agrega embeddings por ID usando diferentes estrategias"""
    df = pd.DataFrame({'id': ids, 'embedding': list(embeddings)})
    
    if aggregation == 'mean':
        aggregated = df.groupby('id')['embedding'].apply(lambda x: np.mean(np.vstack(x), axis=0))
    elif aggregation == 'max':
        aggregated = df.groupby('id')['embedding'].apply(lambda x: np.max(np.vstack(x), axis=0))
    elif aggregation == 'concat':
        # Para concat, limitamos a los primeros 3 elementos para no hacer el vector demasiado grande
        aggregated = df.groupby('id')['embedding'].apply(
            lambda x: np.concatenate([x.iloc[i] for i in range(min(3, len(x)))], axis=0))
    else:
        raise ValueError("Método de agregación no soportado")
    
    return aggregated

# Agregar embeddings de publicaciones por emprendimiento
pub_embeddings_mean = aggregate_text_embeddings(
    contenido_embeddings, 
    publicaciones['id_emprendimiento'], 
    aggregation='mean'
)

# Agregar embeddings de comentarios por publicación y luego por emprendimiento
comentarios['id_emprendimiento'] = comentarios['id_publicacion'].map(
    publicaciones.set_index('id_publicacion')['id_emprendimiento']
)
com_embeddings_mean = aggregate_text_embeddings(
    comentario_embeddings, 
    comentarios['id_emprendimiento'], 
    aggregation='mean'
)
````

````python
# 2. Fusión de características textuales

def combine_text_features(desc_emb, pub_emb, com_emb, method='weighted'):
    """Combina diferentes embeddings textuales"""
    emprendimientos_ids = desc_emb.index.union(pub_emb.index).union(com_emb.index)
    
    combined_features = {}
    for emp_id in emprendimientos_ids:
        features = []
        
        # Embedding de descripción (si existe)
        if emp_id in desc_emb:
            features.append(desc_emb.loc[emp_id])
        
        # Embedding de publicaciones (si existe)
        if emp_id in pub_emb:
            if method == 'weighted':
                # Ponderar más las publicaciones recientes
                pub_weight = min(1.0, len(publicaciones[publicaciones['id_emprendimiento'] == emp_id]) * 0.2)
                features.append(pub_emb.loc[emp_id] * pub_weight)
            else:
                features.append(pub_emb.loc[emp_id])
        
        # Embedding de comentarios (si existe)
        if emp_id in com_emb:
            if method == 'weighted':
                # Ponderar menos los comentarios
                features.append(com_emb.loc[emp_id] * 0.5)
            else:
                features.append(com_emb.loc[emp_id])
        
        # Promedio ponderado de los embeddings disponibles
        if features:
            if method == 'weighted':
                weights = [1.0] + [0.7]*(len(features)-1)  # Más peso a la descripción
                combined = np.average(np.vstack(features), axis=0, weights=weights)
            else:
                combined = np.mean(np.vstack(features), axis=0)
            
            combined_features[emp_id] = combined
    
    return combined_features

# Combinar características textuales
text_features = combine_text_features(
    pd.Series({i: emb for i, emb in enumerate(descripcion_embeddings)}),
    pub_embeddings_mean,
    com_embeddings_mean,
    method='weighted'
)
````

````python
# 3. Reducción de dimensionalidad para características textuales
# (Opcional, pero recomendado para mantener el modelo manejable)

def reduce_dimensions(features_dict, n_components=256):
    """Reduce la dimensionalidad de los embeddings usando SVD"""
    ids = list(features_dict.keys())
    embeddings = np.vstack([features_dict[id] for id in ids])
    
    svd = TruncatedSVD(n_components=n_components)
    reduced_embeddings = svd.fit_transform(embeddings)
    
    return {id: reduced_embeddings[i] for i, id in enumerate(ids)}

text_features_reduced = reduce_dimensions(text_features, n_components=64)
````

````python
# 4. Extracción de palabras clave con TF-IDF (para interpretabilidad)

import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.corpus import stopwords

def extract_keywords(texts, n_keywords=5):
    """Extrae palabras clave usando TF-IDF"""
    # Descargar stopwords en español si no están disponibles
    try:
        spanish_stopwords = stopwords.words('spanish')
    except:
        nltk.download('stopwords')
        spanish_stopwords = stopwords.words('spanish')
    
    # Crear el vectorizador TF-IDF
    tfidf = TfidfVectorizer(
        max_features=1000, 
        stop_words=spanish_stopwords
    )
    
    # Ajustar el vectorizador y transformar los textos
    tfidf_matrix = tfidf.fit_transform(texts.fillna(''))
    
    # Extraer palabras clave para cada texto
    keywords = {}
    for i, text in enumerate(texts):
        if pd.notna(text) and text.strip():
            feature_index = tfidf_matrix[i,:].nonzero()[1]
            tfidf_scores = zip(feature_index, [tfidf_matrix[i, x] for x in feature_index])
            sorted_scores = sorted(tfidf_scores, key=lambda x: x[1], reverse=True)
            keywords[i] = [tfidf.get_feature_names_out()[i] for (i, s) in sorted_scores[:n_keywords]]
        else:
            keywords[i] = []
    
    return keywords

# Extraer palabras clave de descripciones
desc_keywords = extract_keywords(emprendimientos['descripcion'])
````

````python
# 5. Integración con el grafo existente

# Agregar características textuales al grafo
for emp_id, features in text_features_reduced.items():
    if emp_id in G.nodes:
        G.nodes[emp_id]['text_features'] = features

# Agregar palabras clave al grafo (para interpretabilidad)
for emp_id, keywords in desc_keywords.items():
    if emp_id in G.nodes:
        G.nodes[emp_id]['keywords'] = keywords
````

````python
# 6. Fusión final de características (numéricas + textuales)

def combine_all_features(G, text_weight=0.6):
    """Combina características estructurales y textuales"""
    for node in G.nodes():
        if 'features' in G.nodes[node] and 'text_features' in G.nodes[node]:
            # Escalar características numéricas
            numeric_features = G.nodes[node]['features']
            numeric_features = MinMaxScaler().fit_transform(numeric_features.reshape(1, -1)).flatten()
            
            # Combinar ponderadamente
            text_features = G.nodes[node]['text_features']
            combined = np.concatenate([
                numeric_features * (1 - text_weight),
                text_features * text_weight
            ])
            
            G.nodes[node]['combined_features'] = combined
        elif 'features' in G.nodes[node]:
            G.nodes[node]['combined_features'] = G.nodes[node]['features']
        elif 'text_features' in G.nodes[node]:
            G.nodes[node]['combined_features'] = G.nodes[node]['text_features']

combine_all_features(G, text_weight=0.6)
````

````python
# 7. Preparación para GraphSAGE

# Convertir a formato compatible con GraphSAGE
node_features = np.array([
    G.nodes[node]['combined_features'] 
    for node in G.nodes()
    if 'combined_features' in G.nodes[node]
])

edge_list = np.array(G.edges())

print(f"Características finales por nodo: {node_features.shape[1]} dimensiones")
print(f"Número total de nodos con características: {len(node_features)}")
print(f"Número total de aristas: {len(edge_list)}")
````

````python
from torch_geometric.data import Data

# 1. Obtener node_features del grafo G
node_features = []
node_ids = sorted(G.nodes())  # Ordenamos los IDs para mantener consistencia

for node_id in node_ids:
    # Usamos las características combinadas que creamos anteriormente
    features = G.nodes[node_id]['combined_features']
    node_features.append(features)

# Convertir a tensor de PyTorch
node_features = torch.tensor(np.array(node_features), dtype=torch.float)

# 2. Obtener edge_index (conexiones) del grafo G
edge_index = []
edge_weight = []

for src, dst in G.edges():
    # Obtenemos el peso de la arista (usamos 1.0 si no existe)
    weight = G[src][dst].get('weight', 1.0)
    
    # Añadimos ambas direcciones para grafo no dirigido
    edge_index.append([src, dst])
    edge_weight.append(weight)
    
    edge_index.append([dst, src])
    edge_weight.append(weight)

# Convertir a tensores de PyTorch
edge_index = torch.tensor(edge_index, dtype=torch.long).t().contiguous()
edge_weight = torch.tensor(edge_weight, dtype=torch.float)

# 3. Crear el objeto Data
data = Data(
    x=node_features,          # Características de los nodos
    edge_index=edge_index,    # Índices de las aristas
    edge_attr=edge_weight     # Atributos/pesos de las aristas
)

# Opcional: Añadir otros atributos que puedan ser útiles
data.num_nodes = len(node_ids)
data.node_ids = torch.tensor(node_ids, dtype=torch.long)

# Verificación
print("Objeto Data creado:")
print(data)
print(f"Número de nodos: {data.num_nodes}")
print(f"Número de características por nodo: {data.num_node_features}")
print(f"Número de aristas: {data.edge_index.shape[1]}")
````

## Ensayo de Pooling + TF-IDF

````python
def combine_embeddings(embeddings):
    if not embeddings:
        return np.zeros(768)
    stack = np.stack([emb for emb in embeddings if emb is not None])
    mean_pool = stack.mean(axis=0)
    max_pool = stack.max(axis=0)
    return np.concatenate([mean_pool, max_pool])  # output: 768 * 2 = 1536
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# Obtener corpus
corpus_pub = publicaciones['contenido'].fillna('').tolist()
corpus_com = comentarios['comentario'].fillna('').tolist()

# Crear vectorizadores
tfidf_pub = TfidfVectorizer(max_features=100)
tfidf_com = TfidfVectorizer(max_features=100)

# Ajustar y transformar
tfidf_pub_matrix = tfidf_pub.fit_transform(corpus_pub).toarray()
tfidf_com_matrix = tfidf_com.fit_transform(corpus_com).toarray()

# Añadir a los DataFrames
publicaciones['contenido_tfidf'] = list(tfidf_pub_matrix)
comentarios['comentario_tfidf'] = list(tfidf_com_matrix)
````

````python
for emp_id in G.nodes:
    # Embedding de descripción
    desc_embedding = emprendimientos[emprendimientos['id_emprendimiento'] == emp_id]['descripcion_embedding'].iloc[0]

    # Publicaciones
    pubs = publicaciones[publicaciones['id_emprendimiento'] == emp_id]
    pub_embeddings = [emb for emb in pubs['contenido_embedding'] if emb is not None]
    pub_pooled = combine_embeddings(pub_embeddings)

    pub_tfidfs = list(pubs['contenido_tfidf']) if 'contenido_tfidf' in pubs else []
    pub_tfidf_mean = np.mean(pub_tfidfs, axis=0) if pub_tfidfs else np.zeros(100)

    # Comentarios
    pub_ids = pubs['id_publicacion'].tolist()
    coms = comentarios[comentarios['id_publicacion'].isin(pub_ids)]
    com_embeddings = [emb for emb in coms['comentario_embedding'] if emb is not None]
    com_pooled = combine_embeddings(com_embeddings)

    com_tfidfs = list(coms['comentario_tfidf']) if 'comentario_tfidf' in coms else []
    com_tfidf_mean = np.mean(com_tfidfs, axis=0) if com_tfidfs else np.zeros(100)

    # Vector combinado final
    combined_embedding = np.concatenate([
        desc_embedding,        # 768
        pub_pooled,            # 1536
        pub_tfidf_mean,        # 100
        com_pooled,            # 1536
        com_tfidf_mean         # 100
    ])  # total: 4040 dimensiones

    G.nodes[emp_id]['text_features'] = combined_embedding
````

````python
from torch_geometric.data import Data

for emp_id in G.nodes:
    numeric_categ_features = G.nodes[emp_id].get('features', np.zeros(features_df.shape[1]-1))
    text_features = G.nodes[emp_id].get('text_features', np.zeros(4040))  # nuevo tamaño
    combined_features = np.concatenate([numeric_categ_features, text_features])
    G.nodes[emp_id]['combined_features'] = combined_features

# Convertir a tensor PyG
node_features = torch.tensor([G.nodes[n]['combined_features'] for n in G.nodes], dtype=torch.float)
edge_index = torch.tensor([[e[0], e[1]] for e in G.edges], dtype=torch.long).t()
edge_weight = torch.tensor([G[e[0]][e[1]]['weight'] for e in G.edges], dtype=torch.float)

data = Data(x=node_features, edge_index=edge_index, edge_attr=edge_weight)
````

# Combinar Embeddings Por Emprendimiento

## 📌 ¿Qué hace este bloque de código?

Para cada emprendimiento en el grafo `G`, **calcula una representación vectorial (embedding) combinada**, usando:

1. La descripción del emprendimiento.
2. Los contenidos de sus publicaciones.
3. Los comentarios asociados a esas publicaciones.

Este vector se guarda como atributo del nodo bajo la clave `'text_features'`.

---

## 🧠 Paso a paso explicativo

```python
for emp_id in G.nodes:
```

* Recorremos cada **emprendimiento** (nodo) del grafo.

---

### 1. 📄 Embedding de la descripción

```python
desc_embedding = emprendimientos[emprendimientos['id_emprendimiento'] == emp_id]['descripcion_embedding'].iloc[0]
```

* Extrae el embedding de la **descripción textual** del emprendimiento.
* Se espera que ya tengas una columna `'descripcion_embedding'` generada previamente con BETO u otro modelo.

📌 **Ejemplo visual:**

| id\_emprendimiento | descripción                        | descripcion_embedding  |
| ------------------ | ---------------------------------- | ------------------------- |
| 1                  | "Emprendimiento de café artesanal" | `[0.02, -0.01, ..., 0.15]`  |

---

### 2. 📢 Promedio de embeddings de publicaciones

```python
pub_embeddings = publicaciones[publicaciones['id_emprendimiento'] == emp_id]['contenido_embedding']
pub_embedding = np.mean([emb for emb in pub_embeddings if emb is not None], axis=0) if len(pub_embeddings) > 0 else np.zeros(768)
```

* Busca todas las publicaciones del emprendimiento.
* Promedia sus embeddings si existen, o usa un vector cero.

📌 **Ejemplo visual:**

| id\_publicacion | id\_emprendimiento | contenido            | contenido_embedding  |
| --------------- | ------------------ | -------------------- | -------------------------- |
| 10              | 1                  | "Café 100% orgánico" | `[0.1, 0.05, ..., -0.02]`  |
| 11              | 1                  | "Nuevas variedades"  | `[0.08, 0.03, ..., 0.01]`  |

> Se promedian los vectores `[0.1, ...]` y `[0.08, ...]`.

---

### 3. 💬 Promedio de embeddings de comentarios

```python
pub_ids = publicaciones[publicaciones['id_emprendimiento'] == emp_id]['id_publicacion'].tolist()
com_embeddings = comentarios[comentarios['id_publicacion'].isin(pub_ids)]['comentario_embedding']
com_embedding = np.mean([emb for emb in com_embeddings if emb is not None], axis=0) if len(com_embeddings) > 0 else np.zeros(768)
```

* Busca todos los comentarios **sobre las publicaciones** del emprendimiento.
* Promedia sus embeddings.

📌 **Ejemplo visual:**

| id\_comentario | id\_publicacion | comentario            | comentario\_embedding  |
| -------------- | --------------- | --------------------- | -------------------------- |
| 100            | 10              | "Me encantó el sabor" | `[0.05, -0.03, ..., 0.09]` |
| 101            | 11              | "Excelente servicio"  | `[0.07, 0.01, ..., 0.04]`  |

---

### 4. 🧩 Combinar todos los embeddings

```python
combined_embedding = (desc_embedding + pub_embedding + com_embedding) / 3
```

* Se promedian los tres vectores para obtener **una única representación textual** del emprendimiento.

📌 **Visualización conceptual:**

```
combined_embedding = promedio(
    descripción_embedding,
    publicaciones_embedding,
    comentarios_embedding
)
```

> Esto da un vector 768-D con información textual **integrada**.

---

### 5. 📎 Añadirlo al grafo

```python
G.nodes[emp_id]['text_features'] = combined_embedding
```

* Almacena el vector en el nodo del grafo como atributo `'text_features'`.

---

## ✅ ¿Por qué es útil?

* Este vector combinado refleja de forma más **rica** el contenido textual que rodea al emprendimiento.
* Se puede usar como entrada a modelos de aprendizaje sobre grafos, como **GraphSAGE**, **GAT** o **GCN**.
* Permite que nodos sin publicaciones o comentarios aún tengan al menos un embedding basado en la descripción.

---


````python
# Combinar embeddings por emprendimiento
for emp_id in G.nodes:
    # Embedding de la descripción
    desc_embedding = emprendimientos[emprendimientos['id_emprendimiento'] == emp_id]['descripcion_embedding'].iloc[0]

    # Promedio de embeddings de publicaciones
    pub_embeddings = publicaciones[publicaciones['id_emprendimiento'] == emp_id]['contenido_embedding']
    pub_embedding = np.mean([emb for emb in pub_embeddings if emb is not None], axis=0) if len(pub_embeddings) > 0 else np.zeros(768)

    # Promedio de embeddings de comentarios
    pub_ids = publicaciones[publicaciones['id_emprendimiento'] == emp_id]['id_publicacion'].tolist()
    com_embeddings = comentarios[comentarios['id_publicacion'].isin(pub_ids)]['comentario_embedding']
    com_embedding = np.mean([emb for emb in com_embeddings if emb is not None], axis=0) if len(com_embeddings) > 0 else np.zeros(768)

    # Combinar embeddings
    combined_embedding = (desc_embedding + pub_embedding + com_embedding) / 3
    G.nodes[emp_id]['text_features'] = combined_embedding

print("Ejemplo de embedding de texto:", G.nodes[emprendimientos['id_emprendimiento'].iloc[0]]['text_features'][:5])
````

# GraphSAGE

````python
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv
from torch_geometric.loader import DataLoader
from sklearn.metrics import precision_score, f1_score, recall_score, roc_auc_score, average_precision_score
from sklearn.preprocessing import StandardScaler
from tqdm import tqdm
import matplotlib.pyplot as plt
from torch.utils.tensorboard import SummaryWriter

# 1. Configuración de semillas para reproducibilidad
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

class GraphSAGE(nn.Module):
    def __init__(self, in_channels, hidden_channels=512, out_channels=256, num_layers=2, dropout=0.5):
        super().__init__()
        self.convs = nn.ModuleList()
        self.convs.append(SAGEConv(in_channels, hidden_channels))
        for _ in range(num_layers - 2):
            self.convs.append(SAGEConv(hidden_channels, hidden_channels))
        self.convs.append(SAGEConv(hidden_channels, out_channels))
        
        self.dropout = dropout
        self.predictor = nn.Sequential(
            nn.Linear(2 * out_channels, out_channels),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(out_channels, 1)
        )
    
    def forward(self, x, edge_index):
        for conv in self.convs[:-1]:
            x = conv(x, edge_index)
            x = F.relu(x)
            x = F.dropout(x, p=self.dropout, training=self.training)
        x = self.convs[-1](x, edge_index)
        return x
    
    def predict(self, x_src, x_dst):
        h = torch.cat([x_src, x_dst], dim=1)
        return torch.sigmoid(self.predictor(h)).view(-1)

def train(model, data, optimizer, criterion):
    model.train()
    optimizer.zero_grad()
    
    h = model(data.x, data.edge_index)
    
    # Manejo más eficiente de aristas positivas/negativas
    pos_mask = data.edge_label == 1
    pos_pred = model.predict(h[data.edge_label_index[0][pos_mask]],
                           h[data.edge_label_index[1][pos_mask]])
    neg_pred = model.predict(h[data.edge_label_index[0][~pos_mask]],
                           h[data.edge_label_index[1][~pos_mask]])
    
    loss = criterion(pos_pred, torch.ones_like(pos_pred)) + \
           criterion(neg_pred, torch.zeros_like(neg_pred))
    
    loss.backward()
    
    # Clip de gradientes para estabilidad
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    
    optimizer.step()
    return loss.item()

@torch.no_grad()
def test(model, data):
    model.eval()
    h = model(data.x, data.edge_index)
    pred = model.predict(h[data.edge_label_index[0]], h[data.edge_label_index[1]])
    y = data.edge_label.float()
    
    pred_np = pred.cpu().numpy()
    y_np = y.cpu().numpy()
    pred_binary = (pred_np > 0.5).astype(int)
    
    return {
        'auc': roc_auc_score(y_np, pred_np),
        'ap': average_precision_score(y_np, pred_np),
        'precision': precision_score(y_np, pred_binary, zero_division=0),
        'recall': recall_score(y_np, pred_binary, zero_division=0),
        'f1': f1_score(y_np, pred_binary, zero_division=0),
        'loss': F.binary_cross_entropy(pred, y).item()
    }

def train_model(train_data, val_data, test_data, num_epochs=1000, patience=50):
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    # Normalización de características
    scaler = StandardScaler()
    train_data.x = torch.tensor(scaler.fit_transform(train_data.x.cpu()), dtype=torch.float).to(device)
    val_data.x = torch.tensor(scaler.transform(val_data.x.cpu()), dtype=torch.float).to(device)
    test_data.x = torch.tensor(scaler.transform(test_data.x.cpu()), dtype=torch.float).to(device)
    
    # Configuración del modelo
    model = GraphSAGE(
        in_channels=train_data.num_node_features,
        hidden_channels=512,
        out_channels=256,
        num_layers=2,
        dropout=0.5
    ).to(device)
    
    # Optimizador y scheduler
    optimizer = torch.optim.AdamW(model.parameters(), lr=0.0005, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='max', factor=0.5, patience=10, verbose=True
    )
    
    # Función de pérdida con balanceo de clases
    criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([2.0]).to(device))
    
    # TensorBoard
    writer = SummaryWriter()
    
    # Historial de entrenamiento
    history = {
        'train_loss': [],
        'val_metrics': {
            'auc': [], 'ap': [], 'precision': [], 'recall': [], 'f1': [], 'loss': []
        },
        'lr': []
    }
    
    best_val_auc = 0
    best_val_loss = float('inf')
    best_epoch = 0
    no_improve = 0
    
    for epoch in tqdm(range(num_epochs), desc="Entrenando"):
        # Entrenamiento
        train_loss = train(model, train_data, optimizer, criterion)
        history['train_loss'].append(train_loss)
        
        # Validación
        val_metrics = test(model, val_data)
        for metric_name, metric_value in val_metrics.items():
            history['val_metrics'][metric_name].append(metric_value)
        
        # Actualizar LR scheduler
        scheduler.step(val_metrics['auc'])
        history['lr'].append(optimizer.param_groups[0]['lr'])
        
        # Early stopping mejorado
        if val_metrics['auc'] > best_val_auc or val_metrics['loss'] < best_val_loss:
            if val_metrics['auc'] > best_val_auc:
                best_val_auc = val_metrics['auc']
            if val_metrics['loss'] < best_val_loss:
                best_val_loss = val_metrics['loss']
            best_epoch = epoch
            no_improve = 0
            torch.save(model.state_dict(), 'best_model.pth')
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"\nEarly stopping at epoch {epoch}")
                break
        
        # Logging en TensorBoard
        writer.add_scalar('Loss/train', train_loss, epoch)
        writer.add_scalar('Loss/val', val_metrics['loss'], epoch)
        writer.add_scalar('AUC/val', val_metrics['auc'], epoch)
        writer.add_scalar('LR', optimizer.param_groups[0]['lr'], epoch)
        
        # Mostrar métricas
        if epoch % 5 == 0:
            train_metrics = test(model, train_data)
            tqdm.write(
                f"Epoch {epoch:03d}: "
                f"Train Loss: {train_loss:.4f}, "
                f"Precision: {train_metrics['precision']:.4f}, "
                f"F1: {train_metrics['f1']:.4f}, "
                f"AUC: {train_metrics['auc']:.4f}, "
                f"Recall: {train_metrics['recall']:.4f} |\n"
                f"Val Loss: {val_metrics['loss']:.4f}, "
                f"Precision: {val_metrics['precision']:.4f}, "
                f"F1: {val_metrics['f1']:.4f}, "
                f"AUC: {val_metrics['auc']:.4f}, "
                f"Recall: {val_metrics['recall']:.4f}, "
                f"LR: {optimizer.param_groups[0]['lr']:.2e}"
            )
    
    writer.close()
    
    # Cargar mejor modelo y evaluar
    model.load_state_dict(torch.load('best_model.pth'))
    test_metrics = test(model, test_data)
    
    print(f'\nMejor modelo en epoch {best_epoch}:')
    print(
        f"Test Metrics: "
        f"Precision: {test_metrics['precision']:.4f}, "
        f"F1: {test_metrics['f1']:.4f}, "
        f"AUC: {test_metrics['auc']:.4f}, "
        f"Recall: {test_metrics['recall']:.4f}, "
        f"AP: {test_metrics['ap']:.4f}, "
        f"Loss: {test_metrics['loss']:.4f}"
    )
    
    # Visualización del entrenamiento
    plot_training(history)
    
    return model, history, test_metrics, test_data

def plot_training(history):
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.plot(history['train_loss'], label='Train')
    plt.plot(history['val_metrics']['loss'], label='Val')
    plt.title('Loss')
    plt.legend()
    
    plt.subplot(1, 3, 2)
    plt.plot(history['val_metrics']['auc'], label='AUC')
    plt.plot(history['val_metrics']['f1'], label='F1')
    plt.title('Validation Metrics')
    plt.legend()
    
    plt.subplot(1, 3, 3)
    plt.plot(history['lr'], label='Learning Rate')
    plt.title('Learning Rate Schedule')
    plt.legend()
    
    plt.tight_layout()
    plt.show()

# Ejecución del entrenamiento
if __name__ == "__main__":
    # Aquí cargarías tus datos
    # train_data, val_data, test_data = cargar_tus_datos()
    
    model, history, test_metrics, test_data = train_model(
        train_data, val_data, test_data,
        num_epochs=1000,
        patience=50
    )
````

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator

def plot_training_results(history, test_metrics=None, figsize=(15, 10), save_path=None):
    """
    Genera gráficas de las métricas de entrenamiento y validación
    
    Args:
        history (dict): Diccionario con historial de entrenamiento
        test_metrics (tuple): Tupla con (test_auc, test_ap) (opcional)
        figsize (tuple): Tamaño de la figura
        save_path (str): Ruta para guardar la figura (opcional)
    """
    # Configuración del estilo
    plt.style.use('seaborn-v0_8') if 'seaborn-v0_8' in plt.style.available else plt.style.use('ggplot')
    
    # Crear figura
    fig, axs = plt.subplots(2, 2, figsize=figsize)
    fig.suptitle('Resultados del Entrenamiento - GraphSAGE', fontsize=16, y=1.02)
    
    # 1. Gráfico de Pérdida (Loss)
    axs[0, 0].plot(history['train_loss'], label='Pérdida en Entrenamiento', color='royalblue', linewidth=2)
    axs[0, 0].set_title('Evolución de la Pérdida')
    axs[0, 0].set_xlabel('Época')
    axs[0, 0].set_ylabel('Pérdida')
    axs[0, 0].grid(True, linestyle='--', alpha=0.7)
    axs[0, 0].xaxis.set_major_locator(MaxNLocator(integer=True))
    
    # 2. Gráfico de AUC-ROC
    axs[0, 1].plot(history['val_auc'], label='Validación AUC-ROC', color='darkorange', linewidth=2)
    axs[0, 1].set_title('AUC-ROC en Validación')
    axs[0, 1].set_xlabel('Época')
    axs[0, 1].set_ylabel('AUC-ROC')
    axs[0, 1].set_ylim(0, 1.05)
    axs[0, 1].grid(True, linestyle='--', alpha=0.7)
    axs[0, 1].xaxis.set_major_locator(MaxNLocator(integer=True))
    
    # 3. Gráfico de Average Precision
    axs[1, 0].plot(history['val_ap'], label='Validación AP', color='forestgreen', linewidth=2)
    axs[1, 0].set_title('Average Precision en Validación')
    axs[1, 0].set_xlabel('Época')
    axs[1, 0].set_ylabel('Average Precision')
    axs[1, 0].set_ylim(0, 1.05)
    axs[1, 0].grid(True, linestyle='--', alpha=0.7)
    axs[1, 0].xaxis.set_major_locator(MaxNLocator(integer=True))
    
    # 4. Gráfico combinado AUC y AP
    axs[1, 1].plot(history['val_auc'], label='AUC-ROC', color='darkorange', linewidth=2)
    axs[1, 1].plot(history['val_ap'], label='Average Precision', color='forestgreen', linewidth=2)
    axs[1, 1].set_title('Comparación de Métricas')
    axs[1, 1].set_xlabel('Época')
    axs[1, 1].set_ylabel('Score')
    axs[1, 1].set_ylim(0, 1.05)
    axs[1, 1].legend()
    axs[1, 1].grid(True, linestyle='--', alpha=0.7)
    axs[1, 1].xaxis.set_major_locator(MaxNLocator(integer=True))
    
    plt.tight_layout()
    
    # Añadir anotación con los mejores valores
    best_auc_epoch = np.argmax(history['val_auc'])
    best_auc = np.max(history['val_auc'])
    best_ap_epoch = np.argmax(history['val_ap'])
    best_ap = np.max(history['val_ap'])
    
    annotation_text = f"Mejor Val AUC: {best_auc:.4f} (Época {best_auc_epoch})\nMejor Val AP: {best_ap:.4f} (Época {best_ap_epoch})"
    
    if test_metrics:
        test_auc, test_ap = test_metrics
        annotation_text += f"\nTest AUC: {test_auc:.4f} | Test AP: {test_ap:.4f}"
    
    fig.text(0.15, -0.08, annotation_text,
             ha='center', fontsize=12, bbox=dict(facecolor='lightgray', alpha=0.5))
    
    # Guardar figura si se especificó una ruta
    #if save_path:
     #   plt.savefig(save_path, bbox_inches='tight', dpi=300)
    
    plt.show()    

plot_training_results(history, test_metrics=(test_auc, test_ap)) # save_path='training_metrics.png'
````

````python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Función para graficar matriz de confusión
def plot_confusion_matrix(model, test_data, threshold=0.5, save_path=None):
    """
    Genera y muestra una matriz de confusión para las predicciones
    
    Args:
        model: Modelo entrenado
        test_data: Datos de prueba
        threshold: Umbral para clasificación
        save_path: Ruta para guardar la figura (opcional)
    """
    
    model.eval()
    with torch.no_grad():
        h = model(test_data.x, test_data.edge_index)
        pred_probs = model.predict(h[test_data.edge_label_index[0]], 
                                 h[test_data.edge_label_index[1]])
        preds = (pred_probs >= threshold).long().cpu().numpy()
        y_true = test_data.edge_label.cpu().numpy()
    
    cm = confusion_matrix(y_true, preds)
    
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Negativo', 'Positivo'],
                yticklabels=['Negativo', 'Positivo'])
    plt.title(f'Matriz de Confusión (Threshold={threshold})')
    plt.ylabel('Verdaderos')
    plt.xlabel('Predichos')
    
    #if save_path:
    #   plt.savefig(save_path, bbox_inches='tight', dpi=300)
    
    plt.show()
    
plot_confusion_matrix(model, test_data) # save_path='confusion_matrix.png'
````

# Realizar Predicciones

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def get_emprendimiento_info(id_emprendimiento, emprendimientos_df):
    """Obtiene información de un emprendimiento dado su ID"""
    emp = emprendimientos_df[emprendimientos_df['id_emprendimiento'] == id_emprendimiento].iloc[0]
    return {
        'id': id_emprendimiento,
        'nombre': emp['nombre_emprendimiento'],
        'descripcion': emp['descripcion'],
        'municipio': emp['id_municipio_origen'],
        'alcance': emp['id_alcance']
    }

def recommend_emprendimientos(model, data, emprendimientos_df, target_id, top_k=10):
    """
    Recomienda emprendimientos similares a uno dado
    
    Args:
        model: Modelo GraphSAGE entrenado
        data: Datos del grafo (en formato PyG)
        emprendimientos_df: DataFrame con información de emprendimientos
        target_id: ID del emprendimiento objetivo
        top_k: Número de recomendaciones a devolver
    """
    # Verificar dispositivo del modelo
    device = next(model.parameters()).device
    
    # Verificar si el emprendimiento objetivo existe en el DataFrame
    if target_id not in emprendimientos_df['id_emprendimiento'].values:
        print(f"Error: No se encontró el emprendimiento con ID {target_id} en el DataFrame")
        return None
    
    # Mover datos al mismo dispositivo que el modelo
    data = data.to(device)
    
    # Obtener embeddings de todos los nodos
    model.eval()
    with torch.no_grad():
        embeddings = model(data.x, data.edge_index).cpu().numpy()
    
    # Asumimos que el orden de los nodos en data.x corresponde a los IDs ordenados
    # Si no es así, necesitaríamos mapear los índices a los IDs
    node_ids = emprendimientos_df['id_emprendimiento'].values
    
    try:
        target_idx = np.where(node_ids == target_id)[0][0]
    except IndexError:
        print(f"Error: El ID {target_id} no existe en los nodos del grafo")
        return None
    
    # Calcular similitud coseno
    target_embedding = embeddings[target_idx].reshape(1, -1)
    similarities = cosine_similarity(target_embedding, embeddings)[0]
    
    # Ordenar por similitud (excluyendo el propio emprendimiento)
    sorted_indices = np.argsort(similarities)[::-1]
    sorted_indices = [i for i in sorted_indices if node_ids[i] != target_id][:top_k]
    
    # Obtener información de los emprendimientos recomendados
    recommendations = []
    for idx in sorted_indices:
        emp_id = node_ids[idx]
        similarity = similarities[idx]
        emp_info = get_emprendimiento_info(emp_id, emprendimientos_df)
        emp_info['similitud'] = f"{similarity:.3f}"
        recommendations.append(emp_info)
    
    return recommendations

    
def print_recommendations(target_emp, recommendations):
    """Muestra las recomendaciones de forma legible"""
    print(f"\nRecomendaciones para el emprendimiento:")
    print(f"ID: {target_emp['id']}")
    print(f"Nombre: {target_emp['nombre']}")
    print(f"Descripción: {target_emp['descripcion']}")
    print(f"Municipio: {target_emp['municipio']}")
    print(f"Alcance: {target_emp['alcance']}")
    
    print("\nEmprendimientos recomendados:")
    for i, rec in enumerate(recommendations, 1):
        print(f"\n#{i} (Similitud: {rec['similitud']})")
        print(f"ID: {rec['id']}")
        print(f"Nombre: {rec['nombre']}")
        print(f"Descripción: {rec['descripcion']}")
        print(f"Municipio: {rec['municipio']}")
        print(f"Alcance: {rec['alcance']}")

# Ejemplo de uso después del entrenamiento:
# 1. Cargar el dataframe de emprendimientos (debes tenerlo cargado como emprendimientos_df)
# emprendimientos_df = pd.read_csv('ruta/a/tu/archivo.csv')

# 2. Seleccionar un emprendimiento para recomendar
# target_id = 5  # Cambia esto por el ID que quieras probar

# 3. Obtener recomendaciones
# target_emp = get_emprendimiento_info(target_id, emprendimientos_df)
# recommendations = recommend_emprendimientos(model, data, emprendimientos_df, target_id)

# 4. Mostrar resultados
# print_recommendations(target_emp, recommendations)
````

````python
from torch_geometric.nn import GCNConv
import torch.nn.functional as F
import torch_geometric.utils as utils
import networkx as nx
import torch

# Modelo GCN
class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, out_channels)
        self.dropout = 0.3  # Reducido para permitir más aprendizaje
        
        # Inicializar pesos
        torch.nn.init.xavier_uniform_(self.conv1.lin.weight)
        torch.nn.init.xavier_uniform_(self.conv2.lin.weight)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)
        x = self.conv2(x, edge_index)
        return x

    def predict_link(self, z, edge_label_index, method='dot'):
        src, dst = z[edge_label_index[0]], z[edge_label_index[1]]
        return (src * dst).sum(dim=1)  # Solo 'dot' para simplicidad

# Añadir grado y clustering coefficient como características
degree_train = utils.degree(train_data.edge_index[0], num_nodes=train_data.x.shape[0]).unsqueeze(1)
degree_val = utils.degree(val_data.edge_index[0], num_nodes=val_data.x.shape[0]).unsqueeze(1)

# Calcular clustering coefficient
G_train = utils.to_networkx(train_data, to_undirected=True)
G_val = utils.to_networkx(val_data, to_undirected=True)
clustering_train = nx.clustering(G_train)
clustering_val = nx.clustering(G_val)
clustering_train_tensor = torch.tensor([clustering_train[i] for i in range(train_data.x.shape[0])]).unsqueeze(1)
clustering_val_tensor = torch.tensor([clustering_val[i] for i in range(val_data.x.shape[0])]).unsqueeze(1)

train_data.x = torch.cat([train_data.x, degree_train, clustering_train_tensor], dim=1)
val_data.x = torch.cat([val_data.x, degree_val, clustering_val_tensor], dim=1)

# Verificar división de datos
print(f"Nodos en train_data: {train_data.x.shape[0]}")
print(f"Enlaces en train_data: {train_data.edge_index.shape[1]}")
print(f"Enlaces en val_data: {val_data.edge_label_index.shape[1]}")
print(f"Clases en val_data: {val_data.edge_label.sum().item()}/{val_data.edge_label.shape[0]} positivas")

# Configuración
model = GCN(in_channels=train_data.x.shape[1], hidden_channels=64, out_channels=16)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50, eta_min=1e-6)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

train_data = train_data.to(device)
val_data = val_data.to(device)
````

````python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from sklearn.metrics import precision_score, f1_score, roc_auc_score, recall_score
from tqdm import tqdm
import numpy as np

# Función de entrenamiento
def train():
    model.train()
    optimizer.zero_grad()
    z = model(train_data.x, train_data.edge_index)
    out = model.predict_link(z, train_data.edge_label_index)
    loss = F.binary_cross_entropy_with_logits(out, train_data.edge_label)
    loss.backward()
    optimizer.step()
    return loss.item()

# Función de evaluación
def evaluate(data):
    model.eval()
    with torch.no_grad():
        z = model(data.x, data.edge_index)
        out = model.predict_link(z, data.edge_label_index)
        preds = (out > 0).float()
        labels = data.edge_label.cpu().numpy()
        preds = preds.cpu().numpy()
        out_sigmoid = torch.sigmoid(out).cpu().numpy()
        
        loss = F.binary_cross_entropy_with_logits(out, data.edge_label).item()
        precision = precision_score(labels, preds, zero_division=0)
        f1 = f1_score(labels, preds, zero_division=0)
        auc = roc_auc_score(labels, out_sigmoid)
        recall = recall_score(labels, preds, zero_division=0)
        
    return loss, precision, f1, auc, recall

# Bucle de entrenamiento
num_epochs = 1500
best_val_auc = 0
best_model_path = 'best_model.pth'
patience = 20
epochs_no_improve = 0

train_losses, val_losses, train_precisions, val_precisions, train_f1s, val_f1s, train_aucs, val_aucs, train_recalls, val_recalls = [], [], [], [], [], [], [], [], [], []

for epoch in tqdm(range(1, num_epochs + 1), desc="Training"):
    train_loss = train()
    val_loss, val_precision, val_f1, val_auc, val_recall = evaluate(val_data)
    train_loss_eval, train_precision, train_f1, train_auc, train_recall = evaluate(train_data)
    
    train_losses.append(train_loss_eval)
    val_losses.append(val_loss)
    train_precisions.append(train_precision)
    val_precisions.append(val_precision)
    train_f1s.append(train_f1)
    val_f1s.append(val_f1)
    train_aucs.append(train_auc)
    val_aucs.append(val_auc)
    train_recalls.append(train_recall)
    val_recalls.append(val_recall)
    
    if val_auc > best_val_auc:
        best_val_auc = val_auc
        torch.save(model.state_dict(), best_model_path)
        epochs_no_improve = 0
    else:
        epochs_no_improve += 1
    
    scheduler.step()
    
    if epoch % 10 == 0:
        print(f'Epoch: {epoch:03d}')
        print(f'Train Loss: {train_loss_eval:.4f}, Precision: {train_precision:.4f}, F1: {train_f1:.4f}, AUC: {train_auc:.4f}, Recall: {train_recall:.4f}')
        print(f'Val Loss: {val_loss:.4f}, Precision: {val_precision:.4f}, F1: {val_f1:.4f}, AUC: {val_auc:.4f}, Recall: {val_recall:.4f}')
    
    #if epochs_no_improve >= patience:
     #   print(f"Early stopping en época {epoch}")
     #   break

# Cargar el mejor modelo
model.load_state_dict(torch.load(best_model_path))
print(f'Mejor AUC de validación: {best_val_auc:.4f}')
````

````python
import torch
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv
import networkx as nx

# Validar número de nodos
num_nodes = len(G.nodes)
assert num_nodes == 80, f"Se esperaban 80 nodos, pero se encontraron {num_nodes}"

# Crear mapeo de IDs de emprendimientos a índices de nodos
mapping = {node: idx for idx, node in enumerate(G.nodes)}
inverse_mapping = {idx: node for node, idx in mapping.items()}

# Modelo GCN
class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, out_channels)
        self.dropout = 0.3  # Reducido para permitir más aprendizaje
        
        # Inicializar pesos
        torch.nn.init.xavier_uniform_(self.conv1.lin.weight)
        torch.nn.init.xavier_uniform_(self.conv2.lin.weight)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=self.dropout, training=self.training)
        x = self.conv2(x, edge_index)
        return x

    def predict_link(self, z, edge_label_index, method='dot'):
        src, dst = z[edge_label_index[0]], z[edge_label_index[1]]
        return (src * dst).sum(dim=1)  # Solo 'dot' para simplicidad

# Configuración del modelo
model = GCN(in_channels=train_data.x.shape[1], hidden_channels=64, out_channels=16)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
train_data = train_data.to(device)

# Cargar el modelo entrenado
best_model_path = 'best_model.pth'
try:
    model.load_state_dict(torch.load(best_model_path))
    print("Modelo cargado exitosamente")
except RuntimeError as e:
    print(f"Error al cargar el modelo: {e}")
    exit()

model.eval()

# Generar embeddings para todos los nodos
with torch.no_grad():
    z = model(train_data.x, train_data.edge_index)
    assert z.shape[0] == num_nodes, f"Se esperaban {num_nodes} embeddings, pero se encontraron {z.shape[0]}"

def predict_links(node_idx, top_k=5):
    """
    Predice las top_k conexiones más probables para un nodo dado

    Args:
        node_idx: Índice del nodo en el grafo
        top_k: Número de recomendaciones a devolver
    Returns:
        Lista de tuplas (id_original, nombre, similitud)
    """
    if node_idx >= num_nodes:
        raise ValueError(f"Índice de nodo {node_idx} fuera de rango (0 a {num_nodes-1})")

    # Normalizar embeddings
    z_norm = F.normalize(z, p=2, dim=1)
    similarities = torch.matmul(z_norm, z_norm[node_idx].unsqueeze(1)).squeeze()

    # Ordenar por similitud descendente
    sorted_sims, indices = torch.sort(similarities, descending=True)

    # Excluir el propio nodo y obtener los top_k
    top_indices = indices[1:top_k+1]
    top_sims = sorted_sims[1:top_k+1]

    recommendations = []
    for idx, sim in zip(top_indices, top_sims):
        idx = int(idx)
        if idx >= num_nodes:
            continue
        original_id = inverse_mapping.get(idx, None)
        if original_id is None:
            continue
        # Obtener nombre del nodo (suponiendo que G.nodes[original_id] tiene atributo 'nombre' o similar)
        nombre = G.nodes[original_id].get('nombre', f"Emprendimiento ID {original_id}")
        recommendations.append((original_id, nombre, float(sim)))

    return recommendations[:top_k]

def find_similar_emprendimientos(emprendimiento_id, top_k=5):
    """
    Encuentra emprendimientos similares basado en los embeddings

    Args:
        emprendimiento_id: ID del emprendimiento en el grafo
        top_k: Número de recomendaciones a devolver
    Returns:
        Tupla (nombre_query, lista de recomendaciones)
    """
    try:
        node_idx = mapping[emprendimiento_id]
        nombre_query = G.nodes[emprendimiento_id].get('nombre', f"Emprendimiento ID {emprendimiento_id}")
        recommendations = predict_links(node_idx, top_k)
        return nombre_query, recommendations
    except KeyError:
        raise KeyError(f"El emprendimiento con ID {emprendimiento_id} no está en el grafo")

````
